## 2. Setup & Data Load



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_PATH = 'data\Aviation_Data.csv'

df = pd.read_csv(DATA_PATH)

df
```

## 3. Data Understanding 
Inspecting columns, missing values, and sample rows.


```python
if 'df' in globals() and isinstance(df, pd.DataFrame):
    display(df.head())
    missing_counts = df.isnull().sum().sort_values(ascending=False).head(20)
    print("\nTop 20 columns with missing values:\n", missing_counts)
else:
    print("DataFrame 'df' not found or is not a pandas DataFrame.")
```

## 4. Data Cleaning



```python
if 'df' in globals():
    # normalize column names
    df.columns = [c.strip().lower().replace(' ', '_') for c in df.columns]
    # parse date column if present
    if 'event_date' in df.columns:
        df['event_date'] = pd.to_datetime(df['event_date'], errors='coerce')
    # create a simplified column for aircraft type if available
    if 'aircraft_type' in df.columns:
        df['aircraft_type_simple'] = df['aircraft_type'].str.split('/').str[0].str.strip().str.lower()
    print('Sample cleaned columns:')
    print(df.columns.tolist()[:30])
    
```

```python
if 'df' in globals():
    # accidents by aircraft type (top 10)
    if 'aircraft_type_simple' in df.columns:
        top_types = df['aircraft_type_simple'].value_counts().nlargest(10)
        print(top_types)
        ax = top_types.plot.barh(figsize=(8,6))
        ax.set_xlabel('Number of Accidents')
        ax.set_title('Top 10 Aircraft Types by Accident Count')
        plt.gca().invert_yaxis()
        plt.show()
    else:
        print('Column aircraft_type_simple not found — adapt this cell to your dataset.')
```

```python
if 'df' in globals():
    cleaned_path = 'data/Aviation_Data_Final.csv'
    df.to_csv(cleaned_path, index=False)
    print('Cleaned CSV saved as', cleaned_path)
```
